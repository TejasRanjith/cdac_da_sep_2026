### assignment 1
- a temperature sensor send continous reading between 20.0 to 25.0 celsius
- the sensor is connected to a microcontroller
- generate 1000 synthetic sensors reading 
- compute the mean and median of your genrated dataset 
- generate visualization of graph 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generate 1000 temperature readings between 20°C and 25°C
generated_readings = np.random.uniform(20.0, 25.0, 1000)

# Calculate mean and median
mean_temperature = np.mean(generated_readings)
median_temperature = np.median(generated_readings)

print("Generated readings:", generated_readings)
print("Mean:", mean_temperature)
print("Median:", median_temperature)

# Mean and median lines
plt.axhline(mean_temperature, linestyle="--", label=f"Mean = {mean_temperature:.2f}°C")
plt.axhline(median_temperature, linestyle=":", label=f"Median = {median_temperature:.2f}°C")

plt.title("Temperature Sensor Readings")
plt.xlabel("Reading Number")
plt.ylabel("Temperature (°C)")
plt.ylim(20, 25)
plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------------
# 1. Generate Temperature Dataset
# -----------------------------------

generated_readings = np.random.uniform(20.0, 25.0, 1000)

# Calculate mean and median
mean_temperature = np.mean(generated_readings)
median_temperature = np.median(generated_readings)

print("Mean Temperature:", mean_temperature)
print("Median Temperature:", median_temperature)


# -----------------------------------
# 2. Line Plot
# -----------------------------------

plt.figure(figsize=(10, 5))

plt.plot(
    range(1, 1001),
    generated_readings,
    linewidth=1
)

plt.axhline(
    mean_temperature,
    linestyle="--",
    label=f"Mean = {mean_temperature:.2f}°C"
)

plt.axhline(
    median_temperature,
    linestyle=":",
    label=f"Median = {median_temperature:.2f}°C"
)

plt.title("Temperature Sensor Readings")
plt.xlabel("Reading Number")
plt.ylabel("Temperature (°C)")
plt.ylim(20, 25)
plt.legend()
plt.grid(True)

plt.show()


# -----------------------------------
# 3. Histogram
# -----------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    generated_readings,
    bins=20,
    edgecolor="black"
)

plt.axvline(
    mean_temperature,
    linestyle="--",
    label=f"Mean = {mean_temperature:.2f}°C"
)

plt.axvline(
    median_temperature,
    linestyle=":",
    label=f"Median = {median_temperature:.2f}°C"
)

plt.title("Distribution of Temperature Readings")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.legend()
plt.grid(True)

plt.show()


# -----------------------------------
# 4. Box Plot
# -----------------------------------

plt.figure(figsize=(6, 5))

plt.boxplot(
    generated_readings,
    patch_artist=True
)

plt.title("Box Plot of Temperature Readings")
plt.ylabel("Temperature (°C)")
plt.ylim(20, 25)
plt.grid(True)

plt.show()


# -----------------------------------
# 5. Heatmap
# -----------------------------------

# Reshape 1000 readings into 20 × 50
temperature_matrix = generated_readings.reshape(20, 50)

plt.figure(figsize=(12, 5))

plt.imshow(
    temperature_matrix,
    cmap="hot",
    aspect="auto",
    interpolation="nearest"
)

plt.colorbar(label="Temperature (°C)")

plt.title("Heatmap of Temperature Sensor Readings")
plt.xlabel("Reading Group")
plt.ylabel("Reading Block")

plt.show()
```
