### Exam Score
    - 100 Students took the exam.
    - your score 85 out of 100
    - 90 students scored less than 85
    - 9 students scored above 85
    - your percentile is 90th percentile

```python
def percentile(scores,target_score):
    less_count = sum(1 for s in scores if s < target_score)
    percentile = (less_count / len(scores)) * 100
    return percentile

test_scores = [55,70,85,90,100]

result = percentile(test_scores, 85)
```

cyclomatic complexity
**M = E − N + 2P**

where:
- **E** = number of edges in the CFG
- **N** = number of nodes in the CFG
- **P** = number of connected components (typically 1 for a single function)

Alternatively, it can be computed as **1 + number of decision points** (e.g., `if`, `while`, `for`, `case`, `&&`, `||`, `? :`, etc.).

**Interpretation:**
- **1–10**: Low complexity, easy to test and maintain.
- **11–20**: Moderate complexity, higher risk of defects.
- **21–50**: High complexity, difficult to test; consider refactoring.
- **>50**: Very high complexity, high risk of errors; should be refactored.

**Uses:**
- Guides testing effort (basis path testing requires at least M test cases).
- Identifies code that is hard to understand and maintain.
- Helps prioritize refactoring and code reviews.

```python
prices = [1,2,3,4,5,6,7,8]
prices = sorted(prices)

def find_median(data):
    data_len = len(data)
    mid = data_len//2
    if data_len % 2 == 0:
        return (data[mid]+data[mid-1])/2.0
    return float(data[mid])

median_1st = find_median(prices)
# median_1st
q1_prices = [_ for _ in prices if _ < median_1st]
q3_prices = [_ for _ in prices if _ > median_1st]
print("IQR : ", find_median(q3_prices)-find_median(q1_prices))
```

A distribution is skewed if it is not symmetric around its mean, meaning one tail is longer or fatter than the other.
- **Positive skew (right-skewed):** The right tail is longer; most data points are clustered on the left.
- **Negative skew (left-skewed):** The left tail is longer; most data points are clustered on the right.
- **Zero skew:** The distribution is perfectly symmetric (e.g., normal distribution).

Skewness is calculated as the third standardized moment:
$$\text{skewness} = \frac{E[(X - \mu)^3]}{\sigma^3}$$
where $X$ is the random variable, $\mu$ is the mean, and $\sigma$ is the standard deviation.

In practice, sample skewness is computed using:
$$g_1 = \left[ \frac{n}{(n-1)(n-2)} \right] \sum \frac{(x_i - \bar{x})^3}{s^3}$$
where $n$ is the sample size, $\bar{x}$ is the sample mean, and $s$ is the sample standard deviation.

This measure helps identify deviations from normality and is useful in financial analysis, quality control, and data preprocessing.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

salaries = np.array([40,45,50,55,310,3])

mean_val = np.mean(salaries)
median_val = np.median(salaries)
skewness_val = stats.skew(salaries)
print("Mean:",mean_val)
print("Median:",median_val)
print("Skewness:",skewness_val)

# plt.hist(salaries, bins = 10, edgecolor = 'black')
# plt.show()


plt.boxplot(salaries,vert=False)
plt.show()
```
