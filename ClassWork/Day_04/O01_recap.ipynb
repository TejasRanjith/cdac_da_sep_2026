```python
# %pip install numpy matplotlib pandas seaborn pprint
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import pprint
```

```python
chips = np.random.normal(loc=100.0, scale=15.0, size=100000)
true_variance = np.var(chips)
sample_variance = np.var(chips, ddof=1)
print(f"True Variance: {true_variance}")
print(f"Sample Variance: {sample_variance}")
```

```python
# empirical verification of Bessel's correction via monte carlo simulation
sample_size = 10
num_trials = 50000
biased_variance = []
unbiased_variance = []

for _ in range(num_trials):
    sample = np.random.choice(chips, size=sample_size, replace=False)
    biased_variance.append(np.var(sample))
    unbiased_variance.append(np.var(sample, ddof=1))
pp = pprint.PrettyPrinter(indent=4)
print(f"Mean of Biased Variance: {np.mean(biased_variance)}")
print(f"Mean of Unbiased Variance: {np.mean(unbiased_variance)}")

```

biased variance = True variance, unbiased variance = sample variance
