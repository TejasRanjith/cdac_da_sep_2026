```python
import matplotlib.pyplot as plt
import scipy.stats as stats
import numpy as np
import random as rd

mu,sigma = 0.0,1.5
sample_normal = np.random.normal(mu, sigma, 1000)

empirical_mean = np.mean(sample_normal)
empirical_std = np.std(sample_normal)


print("Theoretical mean: ",mu)
print("Theoretical standard deviation: ",sigma)
print("Empirical mean: ",empirical_mean)
print("Empirical standard deviation: ",empirical_std)
print("Sample size: ",len(sample_normal))
print("Sample mean: ",np.mean(sample_normal))
print("Sample standard deviation: ",np.std(sample_normal))

plt.figure(figsize=(9,4.5))
count,bins,ignored = plt.hist(sample_normal,bins=50,density=True, alpha=0.5, color='b', label='Empirical Histogram',edgecolor='white')

pdf_normal = stats.norm.pdf(bins,loc=mu, scale=sigma)
plt.plot(bins, pdf_normal, 'r-', linewidth=2, label='Theoretical PDF $\\mathcal{N}(\mu, \sigma^2)$')
plt.title('Normal Distribution: Empirical Histogram vs. Theoretical PDF', fontsize=16,fontweight='bold')
plt.xlabel('x Value')
plt.ylabel('Probability Density')
plt.tight_layout()
plt.legend()
```

## 2. Continous Uniform Distribution: (`np.random.uniform`)
### Overview & Purpose
- The uniform distribution is a continuous probability distribution that describes a set of values that are equally likely to occur. It is often used to model situations where there is no underlying order or pattern to the values, such as the time between events in a Poisson process or the position of a particle in a
random walk.



```python
a,b = -2.0,5.0
mu = 0.0
sigma = 1.0
samples_uniform  = np.random.uniform(a,b,1000)
theortical_mean = (a+b)/2
theortical_var = (b-a)**2/12
print(f"Interval Bounds: [{a:.2f},{b:.2f}]")
print(f"Theoritical Mean: {theortical_mean:.2f} | Empirical Mean: {samples_uniform.mean():.2f}")
print(f"Theoritical Variance: {theortical_var:.2f} | Empirical Variance: {samples_uniform.var():.2f}")

plt.figure(figsize=(10,6))
plt.hist(samples_uniform, bins= 50, density=True, color='green', alpha=0.5, label='Empirical Histogram',edgecolor='white')
x_axis = np.linspace(a-1.0,b+1.0,1000)
pdf_uniform = stats.uniform.pdf(x_axis, loc = a,scale =  b-a)

plt.plot(x_axis, pdf_uniform, color='red', label='Theoritical PDF $\\mathcal{U}$(a,b)')
plt.title('Continous Uniform Distribution Density & PDF', fontsize = 12, fontweight = 'bold')
plt.xlabel('x value', fontsize = 12)
plt.ylabel('Probability Density', fontsize = 12)
plt.ylim(0, 1.0 / (b-a) + 0.05)
plt.legend()
plt.tight_layout()
plt.show()
```
