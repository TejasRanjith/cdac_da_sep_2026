```python
import math,statistics
import numpy as np


def get_percentile_linear(data,percentile):
    sorted_data = sorted(data)
    n= len(sorted_data)
    index = (n+1) * percentile / 100 -1
    if index < 0:
        return sorted_data[0]
    elif index >= n-1:
        return sorted_data[-1]
    lower_index = math.floor(index)
    upper_index = math.ceil(index)

    fraction = index - lower_index
    return sorted_data[lower_index] + fraction * (sorted_data[upper_index] - sorted_data[lower_index])

scores = [10,50,60,70,80,85,100]

q1_linear = get_percentile_linear(scores,25)
q3_linear = get_percentile_linear(scores,75)
iqr_linear = q3_linear - q1_linear
print(f"Q1 (Linear Interpolation): {q1_linear}")
print(f"Q3 (Linear Interpolation): {q3_linear}")
print(f"IQR (Linear Interpolation): {iqr_linear}")

q1_stats_linear,_,q3_stats_linear  = statistics.quantiles(scores, n=4)
print(f"Q1 (Statistics): {q1_stats_linear}")
print(f"Q3 (Statistics): {q3_stats_linear}")
print(f"IQR (Statistics): {q3_stats_linear - q1_stats_linear}")

q1_weibull = np.percentile(scores, 25)
q3_weibull = np.percentile(scores, 75)
iqr_weibull = q3_weibull - q1_weibull
print(f"Q1 (Weibull): {q1_weibull}")
print(f"Q3 (Weibull): {q3_weibull}")
print(f"IQR (Weibull): {iqr_weibull}")


q1_stats_weibull,_,q3_stats_weibull  = statistics.quantiles(scores, n=4, method='inclusive')
print(f"Q1 (Statistics_Weibull): {q1_stats_weibull}")
print(f"Q3 (Statistics_Weibull): {q3_stats_weibull}")
print(f"IQR (Statistics_Weibull): {q3_stats_weibull - q1_stats_weibull}")

# print all of them in a tabular format using string formating

# Table of results
methods = [
    ("Linear Interpolation", q1_linear, q3_linear, iqr_linear),
    ("Statistics (exclusive)", q1_stats_linear, q3_stats_linear, q3_stats_linear - q1_stats_linear),
    ("Weibull (numpy.percentile)", q1_weibull, q3_weibull, iqr_weibull),
    ("Statistics (inclusive)", q1_stats_weibull, q3_stats_weibull, q3_stats_weibull - q1_stats_weibull)
]

# Header
header = f"{'Method':<30} {'Q1':>10} {'Q3':>10} {'IQR':>10}"
separator = "-" * len(header)
print()
print(header)
print(separator)
for method, q1, q3, iqr in methods:
    print(f"{method:<30} {q1:>10.2f} {q3:>10.2f} {iqr:>10.2f}")
```
